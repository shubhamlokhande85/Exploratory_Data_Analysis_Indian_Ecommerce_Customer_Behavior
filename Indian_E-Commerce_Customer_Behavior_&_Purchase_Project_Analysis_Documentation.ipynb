# Project Overview

This project analyzes Indian e-commerce customer behavior and purchase data to understand business performance and identify opportunities for improvement.

The main areas of analysis are:

- Revenue and sales performance
- Customer purchasing behavior
- Product and category performance
- Payment method preferences
- Cart abandonment
- Customer satisfaction
- Monthly sales trends
- Marketing channel performance


The project follows:

**Data → Cleaning → Analysis → Visualization → KPI → Insights → Business Decisions**

# Business Problem

An e-commerce company wants to understand its customers and improve sales.

This analysis will answer:

- Which products generate the most revenue?
- Which product categories perform best?
- Which payment methods are most preferred?
- How much cart abandonment occurs?
- What are the monthly revenue trends?
- How satisfied are customers?
- Which marketing channels perform better?
- What business improvements can be made?

# Business Insights

- **Revenue:** The business generated **₹10.12M** from 25,000 sessions.
- **Conversion:** Only **22.46%** of sessions resulted in purchases, showing an opportunity to improve conversion.
- **Cart Abandonment:** **42%** of carts were abandoned, making checkout improvement a key opportunity.
- **Customer Satisfaction:** The average rating is **3.95/5**, indicating generally positive customer experience with room for improvement.
- **Products & Categories:** A small number of products/categories contribute strongly to revenue and should receive priority.
- **Marketing:** High-performing marketing channels should be identified and prioritized.
- **Payment:** Popular payment methods should remain easily available to customers.

 # Recommendations

- Reduce cart abandonment by simplifying the checkout process.
- Improve product pages and offers to increase the **22.46% purchase rate**.
- Keep high-revenue products well stocked.
- Focus marketing budget on high-performing channels.
- Maintain popular payment methods.
- Investigate low-rated products to improve customer satisfaction.
- Use monthly sales trends for inventory and marketing planning.

# Conclusion

The analysis shows a healthy **₹10.12M revenue**, but there is a clear opportunity to improve conversion.

The key business issue is the **42% cart abandonment rate** and **22.46% purchase rate**.

### Business Focus:

**Reduce Cart Abandonment → Increase Purchases → Improve Revenue**

Overall, the business should focus on improving the checkout experience, supporting high-performing products and channels, and continuously improving customer satisfaction.
