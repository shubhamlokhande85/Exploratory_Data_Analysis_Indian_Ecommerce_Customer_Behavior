## B. Exploratory Data Analysis and Visualization

## 1. Purchasing  Analysis

```python
df["purchased"].value_counts()
```

```python
# Purchasing in percentage
df["purchased"].value_counts(normalize=True) * 100
```

```python
# purchase status  
plt.figure(figsize=(8,5))
sns.countplot(data=df,x="purchased",palette='Spectral')
plt.title("Purchase Status")
plt.xlabel("Purchased")
plt.ylabel("Number of Sessions")
plt.show()
```

## 2.Total Revenue

```python
df_revenue = df["revenue"].sum()
print("Total Revenue : ", round(df_revenue))
```

## 3.Revenue by Product Category


```python
# To see which product categories generate the most revenue.

product_category_revenue = df.groupby("product_category")["revenue"].sum().sort_values(ascending=False)
product_category_revenue
```

```python
#Revenue by Product Category
plt.figure(figsize=(13,5))
sns.barplot(data=product_category_revenue)
plt.title("Revenue by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()
```

## 4.Top 10 best performing products

```python
top_products =df.groupby("product_id")["revenue"].sum().sort_values(ascending=False).head(10)
top_products
```

```python
# Top 10 best performing products
plt.figure(figsize=(15,5))
sns.barplot(top_products,palette="Set2")
plt.title("Top 10 Products by Revenue")
plt.xlabel("Product ID")
plt.ylabel("Revenue")
plt.show()
```

## 5. Payment Method Analysis

```python
payment = df["payment_method"].value_counts()
payment
```

```python
# most commonly used payment method plot
plt.figure(figsize=(15,5))
sns.barplot(payment,palette="Blues")
plt.title("Payment Method Preference")
plt.xlabel("Payment Method")
plt.ylabel("Number of Transactions")

plt.show()
```

## 6.Cart Abandonment

```python
# cart abandonment 

cart_rate = df["cart_abandoned"].mean() * 100

print("Cart Abandonment Rate:",round(cart_rate, 2),"%")
```

```python
# cart_abandoned plot
plt.figure(figsize=(8,5))
cart_order = (df["cart_abandoned"].value_counts().sort_values().index)
sns.countplot(data=df,x="cart_abandoned",order=cart_order,palette="Oranges")

plt.title("Cart Abandonment")
plt.xlabel("Cart Abandoned")
plt.ylabel("Number of Sessions")

plt.show()
```

## 7.Customer Rating Analysis

```python
# Average rating by customer 
average_rating = df["rating"].mean()
print("Average Rating:",round(average_rating, 2))
```

```python
df["rating"].value_counts().sort_index()
```

```python
# Customer Rating Distribution plot
plt.figure(figsize=(15,5))

sns.countplot(
    data=df,
    x="rating",
    order=sorted(df["rating"].dropna().unique()),
    palette="Greens"
)

plt.title("Customer Rating Distribution")
plt.xlabel("Rating")
plt.ylabel("Number of Customers")
plt.show()
```

## 8.  Monthly Revenue Trend

```python
# Monthly revenue 
monthly_revenue =df.groupby("visit_month")['revenue'].sum()
monthly_revenue = monthly_revenue.sort_index()
print(monthly_revenue)
```

```python
# Monthly revenue plot
monthly_revenue = df.groupby("visit_month")["revenue"].sum()

plt.figure(figsize=(10, 5))

sns.lineplot(data=monthly_revenue,color="red",marker="o")

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.xticks(rotation=45)

plt.tight_layout()
plt.show()

```

### 9. Marketing Channel Revenue

```python
# marketing channel revenue
marketing_channel_revenue = df.groupby("marketing_channel")['revenue'].sum().sort_values(ascending=True)
marketing_channel_revenue 
```

```python
# Creating a revenue by marketing channel plot 
plt.figure(figsize=(15,5))
sns.barplot(marketing_channel_revenue,palette="viridis")
plt.title("Revenue by Marketing Channel")
plt.xlabel("Marketing Channel")
plt.ylabel("Revenue")
plt.show()
```

## 10. Revenue by Device Type

```python
#Grouping data from device_type and revenue 
df.head()
device_revenue =df.groupby("device_type")["revenue"].sum()
device_revenue
```

```python
# Creating revenue by device type plot 
plt.figure(figsize=(10,8))
sns.barplot(device_revenue,color='red')
plt.title("Revenue by Device type")
plt.xlabel("device_type")
plt.ylabel("revenue")
plt.show()
```

## 11. Revenue by User Type

```python
# grouping data from user_type and revenue numerical variable 
user_revenue = df.groupby('user_type')['revenue'].sum()
user_revenue
```

```python
#creating revenue by user type barplot
plt.figure(figsize=(10,5))
sns.barplot(user_revenue)
plt.title("Revenue by User Type")
plt.xlabel("User Type")
plt.ylabel("Revenue")
plt.show()
```

## C. KPI

```python
total_revenue = df["revenue"].sum()

total_customers = df["customer_id"].nunique()

total_sessions = df["session_id"].nunique()

total_purchases = df["purchased"].sum()

average_rating = df["rating"].mean()

cart_abandonment_rate = (df["cart_abandoned"].mean() * 100)

purchase_rate = (df["purchased"].mean() * 100)
```

```python
print("KPIs")

print("Total Revenue:", round(total_revenue, 2))

print("Total Customers:", total_customers)

print("Total Sessions:", total_sessions)

print("Total Purchases:", total_purchases)

print("Average Rating:", round(average_rating, 2))

print("Purchase Rate:",round(purchase_rate, 2),"%")

print("Cart Abandonment Rate:",round(cart_abandonment_rate, 2),"%")
```

## D. Business Insights

- **Revenue:** The business generated **₹10.12M** from 25,000 sessions.
- **Conversion:** Only **22.46%** of sessions resulted in purchases, showing an opportunity to improve conversion.
- **Cart Abandonment:** **42%** of carts were abandoned, making checkout improvement a key opportunity.
- **Customer Satisfaction:** The average rating is **3.95/5**, indicating generally positive customer experience with room for improvement.
- **Products & Categories:** A small number of products/categories contribute strongly to revenue and should receive priority.
- **Marketing:** High-performing marketing channels should be identified and prioritized.
- **Payment:** Popular payment methods should remain easily available to customers.


 ## E. Recommendations

- Reduce cart abandonment by simplifying the checkout process.
- Improve product pages and offers to increase the **22.46% purchase rate**.
- Keep high-revenue products well stocked.
- Focus marketing budget on high-performing channels.
- Maintain popular payment methods.
- Investigate low-rated products to improve customer satisfaction.
- Use monthly sales trends for inventory and marketing planning.



## F. Conclusion

The analysis shows a healthy **₹10.12M revenue**, but there is a clear opportunity to improve conversion.

The key business issue is the **42% cart abandonment rate** and **22.46% purchase rate**.

### Business Focus:

**Reduce Cart Abandonment → Increase Purchases → Improve Revenue**

Overall, the business should focus on improving the checkout experience, supporting high-performing products and channels, and continuously improving customer satisfaction.
